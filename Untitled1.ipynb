```python
import pandas as pd
import os
import re # 정규 표현식(태그 제거)을 위해 import
from google.colab import files # 결과 파일 다운로드용

# ----------------------------------------------------------------------
# ✅ 설정 영역 ✅
# 1. Colab에 업로드한 '원본' 파일의 정확한 이름을 입력하세요.
MAIN_FILE_NAME = "matched_performances_v2.csv" # 🚨 이전 단계에서 생성된 파일

# 2. '원본' 파일에서 공연명이 있는 '컬럼 이름'을 입력하세요.
MAIN_TITLE_COLUMN = "Unnamed: 1"  # 🚨 예: "Unnamed: 1"

# 3. Colab에 업로드한 '제외할 목록' 파일의 정확한 이름을 입력하세요.
EXCLUSION_FILE_NAME = "아동 서울 500석.csv" # 🚨

# 4. '제외할 목록' 파일에서 공연명이 있는 '컬럼 이름'을 입력하세요.
EXCLUSION_TITLE_COLUMN = "공연명"      # 🚨 "아동 서울 500석.csv" 파일은 '공연명'이 맞습니다.
# ----------------------------------------------------------------------


def load_dataframe_auto_encoding(filename, header_row=0):
    """파일을 로드하고 (인코딩 자동 시도), 데이터프레임으로 반환하는 함수"""

    print(f"👉 '{filename}' 파일을 찾습니다...")
    if not os.path.exists(filename):
        print(f"\n[오류] '{filename}' 파일을 찾을 수 없습니다!")
        print("왼쪽 '파일' 탭에 파일이 업로드되었는지, 파일명이 정확한지 확인하세요.")
        return None

    print(f"📄 '{filename}' 파일 찾기 성공!")

    try:
        if filename.endswith('.csv'):
            try:
                df = pd.read_csv(filename, encoding='euc-kr', header=header_row)
            except UnicodeDecodeError:
                try:
                    df = pd.read_csv(filename, encoding='cp949', header=header_row)
                except UnicodeDecodeError:
                    df = pd.read_csv(filename, encoding='utf-8', header=header_row)

        elif filename.endswith('.xlsx') or filename.endswith('.xls'):
            df = pd.read_excel(filename, header=header_row)

        else:
            print(f"[오류] 지원하지 않는 파일 형식입니다.")
            return None

        print(f"'{filename}' 파일 로드 성공.\n")
        return df

    except Exception as e:
        print(f"'{filename}' 파일 로드 중 오류 발생: {e}")
        return None

def filter_performances():

    # --- 1. 파일 로드 ---

    # 원본 파일 로드
    df_main = load_dataframe_auto_encoding(MAIN_FILE_NAME, header_row=0)
    if df_main is None: return

    # [중요] '아동' 파일은 상단 3줄의 불필요한 헤더가 있으므로 header=3 (4번째 줄)로 읽습니다.
    # (이전 파일 분석에서 estimatedRowsAboveHeader=3 임을 확인)
    df_exclude = load_dataframe_auto_encoding(EXCLUSION_FILE_NAME, header_row=3)
    if df_exclude is None: return

    # --- 2. 컬럼명 유효성 검사 ---
    print("--- 파일 컬럼 확인 (설정값과 비교해 보세요) ---")
    if MAIN_TITLE_COLUMN not in df_main.columns:
        print(f"🚨 [오류] 원본 파일에 '{MAIN_TITLE_COLUMN}' 컬럼이 없습니다.")
        print(f"   > 사용 가능한 컬럼: {list(df_main.columns)}")
        return

    if EXCLUSION_TITLE_COLUMN not in df_exclude.columns:
        print(f"🚨 [오류] 제외할 파일에 '{EXCLUSION_TITLE_COLUMN}' 컬럼이 없습니다.")
        print(f"   > 사용 가능한 컬럼 (헤더 4번째 줄 기준): {list(df_exclude.columns)}")
        return

    print("컬럼명 확인 완료.")

    # --- 3. 제외할 목록 (Set) 생성 ---
    print("\n'제외할 공연' 목록을 생성합니다...")

    # [태그] 제거 및 공백 정리 함수
    def clean_title(title_series):
        return (
            title_series.astype(str).fillna('')
            .str.replace(r"\[.*?\]", "", regex=True) # 정규식으로 [태그] 제거
            .str.strip()                             # 양쪽 공백 제거
        )

    # '아동' 파일의 공연명을 깨끗하게 정리
    exclusion_base_titles_series = clean_title(df_exclude[EXCLUSION_TITLE_COLUMN])

    # 비어있지 않은, 고유한 '기본 제목'만 Set(집합)으로 만듭니다 (검색 속도가 빠름).
    exclusion_set = set(
        exclusion_base_titles_series[exclusion_base_titles_series.str.len() > 0].unique()
    )

    print(f"'{EXCLUSION_FILE_NAME}' 파일에서 {len(exclusion_set)}개의 고유한 '기본 공연명'을 찾았습니다.")
    # print(f" (예: {list(exclusion_set)[:3]}...)") # 확인용

    # --- 4. 원본 파일 필터링 ---
    print("\n'원본 파일'에서 '제외할 공연'을 필터링합니다...")

    # 원본 파일의 공연명도 '기본 제목'으로 정리하여 임시 컬럼에 저장
    df_main['base_title_main'] = clean_title(df_main[MAIN_TITLE_COLUMN])

    # 'base_title_main'이 'exclusion_set'에 *포함되지 않은(~)* 행만 선택
    # (isin: ~안에 있니? / ~isin: ~안에 없니?)
    mask_not_in_exclusion = ~df_main['base_title_main'].isin(exclusion_set)

    df_filtered = df_main[mask_not_in_exclusion]

    original_count = len(df_main)
    filtered_count = len(df_filtered) # <-- 수정된 부분: len(df_filtered) 추가

    # 'base_title_main' 컬럼은 비교용이었으므로 결과에서 제거
    df_filtered_output = df_filtered.drop(columns=['base_title_main'])

    # --- 5. 결과 출력 ---
    print("\n--- 5. 필터링 결과 ---")
    print(f"✅ 원본 파일에 총 {original_count} 건의 공연이 있었습니다.")
    print(f"❌ {len(exclusion_set)}개의 제외할 공연 목록 중 {original_count - filtered_count} 건이 필터링되었습니다.")
    print(f"✅ 최종적으로 {filtered_count} 건의 공연이 남았습니다.")

    if filtered_count > 0:
        print("\n필터링된 공연 목록의 일부입니다:")
        print(df_filtered_output.head().to_string()) # 상위 5개 행만 출력
    else:
        print("남은 공연이 없습니다.")

    # --- 6. 결과 파일로 저장 및 다운로드 ---
    output_filename = 'filtered_performances.csv'
    df_filtered_output.to_csv(output_filename, index=False, encoding='utf-8-sig')
    print(f"\n필터링 결과를 '{output_filename}' 파일로 저장했습니다.")

    try:
        files.download(output_filename)
        print(f"'{output_filename}' 파일 다운로드를 시작합니다.")
    except Exception as e:
        print(f"파일 다운로드 중 오류 발생: {e}")

# --- 스크립트 실행 ---
filter_performances()
```
